```python
%%sql
select
    table_schema,
    table_name,
    table_type
from information_schema.tables
where table_schema = 'team_tro_sam_ser_pas_stg'
order by table_type, table_name;
```

```python
%%sql
drop table if exists team_tro_sam_ser_pas_stg.flights_clean;
drop table if exists team_tro_sam_ser_pas_stg.airports_clean;
```

```python
%%sql
select 'ods_flights_raw' as object_name, count(*) from team_tro_sam_ser_pas_ods.flights_raw
union all
select 'stg_flights_clean', count(*) from team_tro_sam_ser_pas_stg.flights_clean
union all
select 'stg_flights_deduplicated', count(*) from team_tro_sam_ser_pas_stg.flights_deduplicated
union all
select 'stg_flights_success_raw', count(*) from team_tro_sam_ser_pas_stg.flights_success_raw
union all
select 'stg_flights_cancelled_raw', count(*) from team_tro_sam_ser_pas_stg.flights_cancelled_raw;
```

```python
%%sql
select 'airports_clean' as object_name, count(*) as cnt
from team_tro_sam_ser_pas_stg.airports_clean

union all

select 'airports_deduplicated' as object_name, count(*) as cnt
from team_tro_sam_ser_pas_stg.airports_deduplicated;
```

```python
%%sql
select *
from team_tro_sam_ser_pas_stg.flights_clean
limit 20;
```

```python
%%sql
select
    cancelled,
    count(*) as cnt
from team_tro_sam_ser_pas_stg.flights_clean
group by cancelled
order by cancelled;
```

```python
%%sql
select
    'success' as table_name,
    min(cancelled) as min_cancelled,
    max(cancelled) as max_cancelled,
    count(*) as cnt
from team_tro_sam_ser_pas_stg.flights_success_raw

union all

select
    'cancelled' as table_name,
    min(cancelled) as min_cancelled,
    max(cancelled) as max_cancelled,
    count(*) as cnt
from team_tro_sam_ser_pas_stg.flights_cancelled_raw;
```

```python
%%sql
select
    flight_dt,
    carrier_code,
    carrier_flight_num,
    origin_code,
    dest_code,
    scheduled_dep_tm,
    count(*) as cnt
from team_tro_sam_ser_pas_stg.flights_clean
group by
    flight_dt,
    carrier_code,
    carrier_flight_num,
    origin_code,
    dest_code,
    scheduled_dep_tm
having count(*) > 1
order by cnt desc
limit 20;
```
